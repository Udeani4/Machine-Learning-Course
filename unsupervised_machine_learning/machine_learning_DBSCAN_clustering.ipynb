# DBSCAN Clustering

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
## Create cluster data
X,y=make_moons(n_samples=250,noise=0.05)
```

```python
plt.scatter(X[:,0],X[:,1])
```

```python
## lets add color the identify the categories
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
## feature scaling
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_scaled=scaler.fit_transform(X)
X_scaled
```

```python
dbscan=DBSCAN(eps=0.3)
```

```python
dbscan.fit(X_scaled)
```

```python
dbscan.labels_ ## These are the predicted training labels
```

```python
## lets view how well it classified the clusters
plt.scatter(X[:,0],X[:,1],c=dbscan.labels_)
```

Excellent separation of clusters. You can see that non is miscolored

## Anomaly Detection

We will use DBSCAN to focus on detecting the ouliers in the data.

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.datasets import make_circles
```

```python
X,y=make_circles(n_samples=750,factor=0.3,noise=0.1)
```

```python
X
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
from sklearn.cluster import DBSCAN
```

```python
dbscan=DBSCAN(eps=0.10)
```

```python
dbscan.fit_predict(X) ## use fit_predict to fit and predict to save time
```

You can see all the cluster groups. Wherever we see -1, it is an outlier.

```python
dbscan.labels_
```

```python
## Lets see how the clusters are really represented

plt.scatter(X[:,0],X[:,1],c=dbscan.labels_)
```

Now let us identify the outliers

```python
## Lets pull out the index of the labels that are less than 0
import numpy as np
index = np.where(dbscan.labels_ < 0)
index
```
