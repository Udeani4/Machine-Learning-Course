### CLassification

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
from sklearn.datasets import load_iris

iris=load_iris()
iris
```

```python
iris.DESCR
```

```python
## get independent features
X_data= pd.DataFrame(iris['data'],columns= iris['feature_names'])
X_data.head()
```

```python
## get dependent features
y_data=iris['target']
```

```python
## Train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_data,y_data,test_size=0.2,random_state=10)
```

```python
## apply Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier
dtclassifier=DecisionTreeClassifier()
```

```python
dtclassifier.fit(X_train,y_train)
```

```python
## Visualize the decision tree
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(dtclassifier,filled=True)
```

Looking at the tree we can see at the second level there is a strong difference in the classification. On the left [0,33,1], on the right [0,4,42]. So therefore there is no need to furthere split. If we split too much till we have too many leaf nodes we risk OVERFITTING. Therefore we can:
* Preprune: By tuning the hyperparameter max_features, max_depth, split
* Postprune: start pruning once we have fully developed the tree

With what we have seen we can preprune by setting the max_depth parameter to 2

```python
## set max_depth to 2 to avoid overfitting
dtclassifier2=DecisionTreeClassifier(max_depth=2)
dtclassifier2.fit(X_train,y_train)

plt.figure(figsize=(15,10))
tree.plot_tree(dtclassifier2,filled=True)
```

```python
## Lets compare both performances
## First classifier

from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
y_pred=dtclassifier.predict(X_test)

print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
## Second classifier

y_pred2=dtclassifier2.predict(X_test)

print(accuracy_score(y_test,y_pred2))
print(confusion_matrix(y_test,y_pred2))
print(classification_report(y_test,y_pred2))
```

The pruned classifier has a lower accuracy but it comes in handy to avoid overfitting of the model

NOTE:
Post pruning is good for smaller dataset. Prepruning is more for larger dataset

## Decision Tree Classifier Preruning And Hyperparameter Tuning

To do the pruning we have to play with a lot of parameters.

```python
# dtclassifier3=DecisionTreeClassifier()
```

```python
params = {
    'criterion':["gini","entropy","log_loss"],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5],
    'max_features':['auto','sqrt','log2']  
}
```

```python
from sklearn.model_selection import GridSearchCV

treemodel=DecisionTreeClassifier()

grid=GridSearchCV(estimator=treemodel,param_grid=params,cv=5,scoring='accuracy')
```

```python
import warnings
warnings.filterwarnings('ignore')
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred=grid.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

## Regression

### Diabetes Prediction Using Decision Tree Regressor

```python
from sklearn.datasets import load_diabetes

dataset=load_diabetes()

print(dataset['DESCR'])
```

```python
dataset
```

```python
X_data=pd.DataFrame(dataset['data'],columns=['age','sex','bmi','bp','s1','s2','s3','s4','s5','s6'])
X_data.head()
```

```python
y_data=dataset['target']
```

```python
X_data.shape, y_data.shape
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_data,y_data,test_size=0.3,random_state=10)
```

```python
## correlation
X_train.corr()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(15,10))
sns.heatmap(X_train.corr(),annot=True)
```

```python
from sklearn.tree import DecisionTreeRegressor
dtregressor=DecisionTreeRegressor()
```

```python
dtregressor.fit(X_train,y_train)
```

```python
params={
    'criterion':['square_error','friedman_mse','absolute_error'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5,10,15,20,25],
    'max_features':['auto','sqrt','log2']
}
```

```python
## Hyperparameter Tuning
from sklearn.model_selection import GridSearchCV 
```

```python
grid2=GridSearchCV(estimator=dtregressor,param_grid=params,cv=5,scoring='neg_mean_squared_error')
```

```python
import warnings
warnings.filterwarnings('ignore')
grid2.fit(X_train,y_train)
```

```python
grid2.best_params_
```

```python
grid2.best_score_
```

```python
y_pred=grid2.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```

Very Poor.

```python
## Lets view the tree
## pick the best parameters and insert into the Decision tree object
selectedmodel=DecisionTreeRegressor(criterion='friedman_mse',max_depth=4,max_features='sqrt',splitter= 'random')
```

```python
selectedmodel.fit(X_train,y_train)
```

```python
import matplotlib.pyplot as plt
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(selectedmodel,filled=True)
```
