# K Nearest Neighbours

### Classification Problem (K_Nearest_Neighbour)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline 
```

```python
from sklearn.datasets import make_classification

X,y=make_classification(
    n_samples=1000, #1000 observations
    n_features=3, # 3 total features
    n_redundant=1,
    n_classes=2, #binary target/label
    random_state=999
)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.33,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier

k_classifier=KNeighborsClassifier(n_neighbors=5,algorithm='auto')
k_classifier.fit(X_train,y_train)
```

```python
y_pred=k_classifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report

print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

### Regression Problem (K_Nearest_Regressor)

```python
from sklearn.datasets import make_regression
X,y=make_regression(n_samples=1000,n_features=2,noise=10,random_state=42)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.33,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsRegressor
```

```python
k_regressor=KNeighborsRegressor(n_neighbors=6,algorithm='auto')
k_regressor.fit(X_train,y_train)
```

```python
y_pred=k_regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```

## Task: Finding the best k value using 

```python
X,y=make_classification(
    n_samples=1000, #1000 observations
    n_features=3, # 3 total features
    n_redundant=1,
    n_classes=2, #binary target/label
    random_state=999
)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.33,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_classifier=KNeighborsClassifier()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV

# Define the parameter grid
param_grid = {
    'n_neighbors': list(range(1, 11))   # k values from 1 to 10
}

# Create GridSearchCV object
grid_search = GridSearchCV(
    estimator=knn_classifier,
    param_grid=param_grid,
    cv=5,              # 5-fold cross validation
    scoring='accuracy'
)

# Train the model
grid_search.fit(X_train, y_train)

# Best k value
print("Best k:", grid_search.best_params_)

# Best score
print("Best accuracy:", grid_search.best_score_)
```

```python
y_pred=grid_search.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report

print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

There is an improvement in the classification with k value of 9
